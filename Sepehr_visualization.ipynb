## Part 3 - Slide 45

Consider the data *iris*

1. Create a scatterplot of Sepal.Length vs. Sepal.Width

```python
plot = ggplot(iris, aes(x = "Sepal.Length", y = "Sepal.Width")) + geom_point()
```

2. Label the axes as "Sepal length" and "Sepal width"

```python
plot + labs(x = 'Sepal Length', y = 'Sepal Width')
```

3. Set the limits of the axes to 1..10

```python
ggplot(iris, aes(x = 'Sepal.Length', y = 'Sepal.Width')) + geom_point() + labs(x = 'Sepal Length', y = 'Sepal Width') + xlim(1,10) + ylim(1,10)
```
