## Part 1 - Slide 15

Consider the data *iris*

1. Create a histogram of Sepal.Width


```python
plt.figure(figsize=(6, 4))
plt.hist(iris['Sepal.Width'], color='red', edgecolor='blue')
plt.xlabel('Sepal.Width (cm)')
plt.ylabel('Frequency')
plt.title('Histogram of Iris Sepal.Width')
plt.show()
```

2. Try different numbers of bins

```python
fig, axes = plt.subplots(1,4, figsize=(16,4))
for ax, bins in zip(axes, [5,10,20,40]):
  ax.hist(iris['Sepal.Width'], bins=bins, color='blue', edgecolor='black')
  ax.set_title(f'bins = {bins}')
  ax.set_xlabel('Sepal.Width (cm)')
axes[0].set_ylabel('Frequency')
plt.tight_layout()
plt.show()
```
